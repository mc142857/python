```python
# ===== 任务1 =====
print("===== 任务1：判断回文 =====")
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print("tacocat 是否回文：", is_palindrome("tacocat"))
print("T A C O cat 是否回文：", is_palindrome("T A C O cat"))
print("hello 是否回文：", is_palindrome("hello"))
print()

# ===== 任务2 =====
print("===== 任务2：判断字母顺序 =====")
def is_abecedarian(s):
    s = s.replace(" ", "")
    return all(s[i] <= s[i+1] for i in range(len(s)-1))

print("adder 是否字母顺序：", is_abecedarian("adder"))
print("abcd efgh xyz 是否字母顺序：", is_abecedarian("abcd efgh xyz"))
print("dog 是否字母顺序：", is_abecedarian("dog"))
print()

# ===== 任务3 =====
print("===== 任务3：去掉元音字母 =====")
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return "".join(c for c in s if c not in vowels)

print("remove_vowels('cat') =", remove_vowels("cat"))
print("remove_vowels('audio') =", remove_vowels("audio"))
print("remove_vowels('AEIOU') =", remove_vowels("AEIOU"))
```

```python
# ===== 任务1 =====
print("===== 任务1：列表反转 =====")
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

print("[1, 2, 3] 反转后：", list_reverse([1, 2, 3]))
print("['a', 'b', 'c'] 反转后：", list_reverse(['a', 'b', 'c']))
print()

# ===== 任务2 =====
print("===== 任务2：判断序列是否升序 =====")
def is_sorted(seq):
    return all(seq[i] <= seq[i+1] for i in range(len(seq)-1))

print("[1, 2, 3, 4] 是否升序：", is_sorted([1, 2, 3, 4]))
print("[1, 3, 2, 4] 是否升序：", is_sorted([1, 3, 2, 4]))
print("'abcd' 是否升序：", is_sorted("abcd"))
```

```python
# ===== 任务1 =====
print("===== 任务1：向量数乘 =====")
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("第二个参数必须是整数或浮点数")
    return tuple(x * s for x in t)

print("(1, 2, 3) * 2 =", vec_scalar_mult((1, 2, 3), 2))
print("(1.5, 2.5) * 3 =", vec_scalar_mult((1.5, 2.5), 3))
print()

# ===== 任务2 =====
print("===== 任务2：向量内积 =====")
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度必须一致")
    return float(sum(a * b for a, b in zip(x, y)))

print("(1, 2, 3) · (4, 5, 6) =", vec_inner_product((1, 2, 3), (4, 5, 6)))
print()

# ===== 任务3 =====
print("===== 任务3：检查有效矩阵 =====")
def check_valid_mx(m):
    if not isinstance(m, tuple):
        return False
    if len(m) == 0:
        return False
    row_len = len(m[0])
    for row in m:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_len:
            return False
        for x in row:
            if not isinstance(x, (int, float)):
                return False
    return True

print("((1,2),(3,4)) 是否有效矩阵：", check_valid_mx(((1, 2), (3, 4))))
print("((1,2),(3)) 是否有效矩阵：", check_valid_mx(((1, 2), (3,))))
print()

# ===== 任务4 =====
print("===== 任务4：矩阵向量乘法 =====")
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")
    if len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度不一致")
    return tuple(sum(row[j] * v[j] for j in range(len(v))) for row in mx)

print("[[1,2],[3,4]] * (5,6) =", mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
```

```python
# ===== 任务1 =====
print("===== 任务1：定义Vector类 =====")
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        self.dim = dim
        self.elements = list(elements)

v1 = Vector(3, [1, 2, 3])
v2 = Vector(2)
print("v1 =", v1.elements, " 维度 =", v1.dim)
print("v2 =", v2.elements, " 维度 =", v2.dim)
print()

# ===== 任务2 =====
print("===== 任务2：类属性vs实例属性 =====")
print("选择实例属性。因为每个Vector对象有各自不同的维度和元素，")
print("如果设为类属性会被所有向量共享，不符合向量各自独立的语义。")
print()

# ===== 任务3 =====
print("===== 任务3：比较运算符 =====")
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        self.dim = dim
        self.elements = list(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不能比较")
        return self.elements < other.elements

    def __le__(self, other):
        return self == other or self < other

    def __gt__(self, other):
        return not self <= other

    def __ge__(self, other):
        return not self < other

a = Vector(2, [2, 4])
b = Vector(2, [2, 5])
print("a =", a.elements, " b =", b.elements)
print("a == b：", a == b)
print("a < b：", a < b)
print()

# ===== 任务4 =====
print("===== 任务4：内积dot方法 =====")
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        self.dim = dim
        self.elements = list(elements)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致")
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

v3 = Vector(3, [1, 2, 3])
v4 = Vector(3, [4, 5, 6])
print("v3·v4 =", v3.dot(v4))
print()

# ===== 任务5 =====
print("===== 任务5：乘法运算符 =====")
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        self.dim = dim
        self.elements = list(elements)

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, [x * other for x in self.elements])
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致")
            return Vector(self.dim, [x * y for x, y in zip(self.elements, other.elements)])
        else:
            raise TypeError("不支持的乘法类型")

    def __rmul__(self, other):
        return self * other

v5 = Vector(2, [2, 3])
v6 = Vector(2, [4, 5])
print("2 * v5 =", (2 * v5).elements)
print("v5 * 2 =", (v5 * 2).elements)
print("v5 * v6 =", (v5 * v6).elements)
```
