```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(k):
    if k == 0:
        return 0
    total = 0.0
    factorial = 1
    for i in range(k):
        if i > 0:
            factorial *= i
        total += 1 / factorial
    return total
print(euler_infinite_sum(4))
print(euler_infinite_sum(1))
print(euler_infinite_sum(0))
print()

def euler_approx(epsilon):
    total = 0.0
    factorial = 1
    i = 0
    while True:
        if i > 0:
            factorial *= i
        term = 1 / factorial
        if term < epsilon:
            break
        total += term
        i += 1
    return total
print(euler_approx(0.001))
print()

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
print_euler_sum_table(4)
print()
```

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2
    return n == 1
print(is_power_of_2(8))
print(is_power_of_2(12))
print()

def is_power(b, n):
    while n % b == 0:
        n = n // b
    return n == 1
print(is_power(2, 16))
print(is_power(3, 10))
```
