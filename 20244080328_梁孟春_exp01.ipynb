```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在0到100之间")
    elif score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"
print(score_level(95))
print(score_level(72))
print()


sum_for = 0
for i in range(1, 101):
    sum_for += i
print(sum_for)
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(sum_while)
print()


def factorial_rec(n):
    if n <= 1:
        return 1
    return n * factorial_rec(n - 1)
def factorial_iter(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
print(factorial_rec(0), factorial_iter(0))
print(factorial_rec(5), factorial_iter(5))
print(factorial_rec(10), factorial_iter(10))
print()


def summary(values):
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)
result = summary([1, 2, 3, 4, 5])
print(result)
min_v, max_v, avg_v = result
print(min_v, max_v, avg_v)
print()


def sum_all(*args):
    return sum(args)
print(sum_all())
print(sum_all(5))
print(sum_all(1, 2, 3, 4, 5))
```

```python
import os
print(os.getcwd())
data_dir = os.path.join("exp01_data")
os.makedirs(data_dir, exist_ok=True)


lines = [
    "name,score",
    "Alice,85",
    "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105",
    "Incomplete"
]
filepath = os.path.join(data_dir, "students.txt")
with open(filepath, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))


def parse_student(line):
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量错误")
    name = parts[0].strip()
    score = int(parts[1].strip())
    if name == "":
        raise ValueError("姓名为空")
    if score < 0 or score > 100:
        raise ValueError("成绩超出范围")
    return {"name": name, "score": score}


valid_records = []
invalid_count = 0
with open(filepath, "r", encoding="utf-8") as f:
    all_lines = f.readlines()

for i, line in enumerate(all_lines[1:], 2):
    line = line.strip()
    if not line:
        continue
    try:
        record = parse_student(line)
        valid_records.append(record)
    except ValueError as e:
        invalid_count += 1
        print(f"第{i}行错误: {e}")
print()


total = len(valid_records)
avg = sum(r["score"] for r in valid_records) / total
pass_rate = len([r for r in valid_records if r["score"] >= 60]) / total * 100
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {total}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg:.2f}\n")
    f.write(f"及格率: {pass_rate:.1f}%\n")
with open(report_path, "a", encoding="utf-8") as f:
    f.write("a模式效果\n")


import pickle
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)
print("pickle读取一致:", loaded == valid_records)
print()


from student_utils import parse_student
print(parse_student("Alice,85"))
try:
    parse_student("Grace,abc")
except ValueError as e:
    print(e)
print()


try:
    with open("not_exist.txt", "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError:
    print("文件不存在")
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >= 60
    def __str__(self):
        return f"{self.name}: {self.score}"


students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    print(s, "及格" if s.is_passed() else "不及格")
print()


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        return f"{self.name}: {self.score}, 研究方向: {self.research_topic}"
g = GraduateStudent("Bob", 58, "python")
print(g)
print()


import copy
original = {"name": "Alice", "scores": [85, 90]}
ref = original
ref["scores"][0] = 100
print("直接赋值后原对象:", original["scores"])
original2 = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original2)
shallow["scores"][0] = 100
print("浅拷贝后原对象:", original2["scores"])
original3 = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original3)
deep["scores"][0] = 100
print("深拷贝后原对象:", original3["scores"])
```
