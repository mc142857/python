```python
import re

print("任务1：数字、空白、cat 单词匹配")
text1 = "cat on mat; later: cats and dogs"
print(re.findall(r"\d+", text1))
print(re.findall(r"\s+", text1))
print(re.findall(r"\bcat\b", text1))
print()


print("任务2：交替匹配 cat|dog 与 a|aa|aaa")
text2 = "cat dog bird CAT"
print(re.findall(r"cat|dog", text2))
print(re.findall(r"cat|dog", text2, re.IGNORECASE))
print(re.findall(r"a|aa|aaa", "aaaa"))
print(re.findall(r"aaa|aa|a", "aaaa"))
print()


print("任务3：量词匹配")
text3 = "ct cat caat caaat"
print(re.findall(r"ca*t", text3))
print(re.findall(r"ca+t", text3))
print(re.findall(r"ca?t", text3))
print(re.findall(r"ca{2}t", text3))
print(re.findall(r"ca{1,2}t", text3))
```

```python
import re

print("任务1：提取邮箱用户名和域名")
text1 = "联系：alice@example.com；备选：bob@test.org"
emails = re.findall(r"(\w+)@(\w+\.\w+)", text1)
print(emails)
print()


print("任务2：回溯引用查找连续重复单词")
text2 = "cat cat, dog cat, bird bird"
print(re.findall(r"\b(\w+)\s+\1\b", text2))
print()


print("任务3：日期从年-月-日转换为月-日-年")
text3 = "2024-05-01\n2024/05/02\n2024.05.03"
result3 = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", text3)
print("转换前：")
print(text3)
print("转换后：")
print(result3)
print()


print("任务4：电话号码统一为 (123) 456-7890")
text4 = "联系我：123-456-7890 或 987.654.3210"
result4 = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text4)
print(result4)
```

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

print(soup.title.string)
print(soup.h1.name)
print(soup.h1.get_text())

print()

for p in soup.find_all("p", class_="course"):
    print(p.get_text())
for a in soup.find_all("a"):
    print(a.get_text(), a["href"])

print()

print(soup.get_text())
print(soup.find("p", class_="course").get_text())
print(soup.find("a").get_text())
```
