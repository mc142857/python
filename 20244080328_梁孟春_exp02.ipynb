```python
# ===== 任务1 =====
print("===== 任务1：不同类型的ndarray =====")
arr_int = np.array([1, 2, 3, 4])
arr_float = np.array([1.0, 2.0, 3.0, 4.0])
arr_bool = np.array([True, False, True, False])
print("整数数组：", arr_int)
print("  dtype =", arr_int.dtype, " shape =", arr_int.shape, " ndim =", arr_int.ndim)
print("浮点数组：", arr_float)
print("  dtype =", arr_float.dtype, " shape =", arr_float.shape, " ndim =", arr_float.ndim)
print("布尔数组：", arr_bool)
print("  dtype =", arr_bool.dtype, " shape =", arr_bool.shape, " ndim =", arr_bool.ndim)
print("说明：NumPy数组元素类型统一、连续存储，比Python列表（可混存、分散存储）运算效率高")
print()

# ===== 任务2 =====
print("===== 任务2：reshape改变形状 =====")
a = np.arange(12)
print("原一维数组：", a)
a34 = a.reshape(3, 4)
print("变为(3,4)后：\n", a34)
a232 = a.reshape(2, 3, 2)
print("变为(2,3,2)后：\n", a232)
print("说明：reshape要求新形状各维度乘积必须等于原数组元素总数（这里都是12）")
print()

# ===== 任务3 =====
print("===== 任务3：索引操作 =====")
b = np.arange(12).reshape(3, 4)
print("原数组(3,4)：\n", b)
row2 = b[1, :]
print("取第2行：", row2, " 形状：", row2.shape)
cols13 = b[:, 0:3]
print("取第1至第3列：\n", cols13, " 形状：", cols13.shape)
gt6 = b[b > 6]
print("取大于6的元素：", gt6, " 形状：", gt6.shape)
print()

# ===== 任务4 =====
print("===== 任务4：条件筛选 =====")
c = np.arange(10)
print("原数组：", c)
print("小于2或大于7：", c[(c < 2) | (c > 7)])
print("大于3且小于7：", c[(c > 3) & (c < 7)])
print()

# ===== 任务5 =====
print("===== 任务5：逐元素乘法 vs 矩阵乘法 =====")
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print("m1 =\n", m1)
print("m2 =\n", m2)
print("逐元素乘法 m1*m2：\n", m1 * m2)
print("矩阵乘法 m1@m2：\n", m1 @ m2)
print("说明：*是对应位置直接相乘；@是线性代数的矩阵乘法（行列点积求和）")
```

```python
# ===== 任务1 =====
print("===== 任务1：生成模拟成绩并统计 =====")
np.random.seed(42)
scores = np.random.normal(70, 10, 1000)
scores = np.clip(scores, 0, 100)
print("样本数：", len(scores))
print("最小值：", scores.min())
print("最大值：", scores.max())
print("均值：", scores.mean())
print("标准差：", scores.std())
print("极差：", scores.max() - scores.min())
print()

# ===== 任务2 =====
print("===== 任务2：不放回抽样对比 =====")
arr = np.arange(1, 11)
print("原数组：", arr)
uniform_sample = np.random.choice(arr, 5, replace=False)
print("均匀抽样结果：", uniform_sample)
weights = arr / arr.sum()
weighted_sample = np.random.choice(arr, 5, replace=False, p=weights)
print("按大小加权抽样结果：", weighted_sample)
print()

# ===== 任务3 =====
print("===== 任务3：随机排列 vs 原地打乱 =====")
a = np.arange(10)
print("原数组：", a)
b = np.random.permutation(a)
print("permutation后新数组：", b)
print("permutation后原数组：", a)
np.random.shuffle(a)
print("shuffle后原数组：", a)
print()

# ===== 任务4 =====
print("===== 任务4：缺失值的影响 =====")
scores_nan = scores.copy()
scores_nan[[10, 20, 30]] = np.nan
print("普通均值：", np.mean(scores_nan))
print("忽略缺失值后均值：", np.nanmean(scores_nan))
print("缺失值个数：", np.isnan(scores_nan).sum())
print("说明：含nan时普通统计函数会返回nan；用nanmean等函数可忽略缺失值计算")
print()

# ===== 任务5 =====
print("===== 任务5：整理表格变量 =====")
table = {'序号': np.arange(1000), '成绩': scores_nan}
print("表格变量已生成，共", len(table['序号']), "条记录")
print("前5条序号与成绩：", table['序号'][:5], table['成绩'][:5])
```

```python
import warnings
warnings.filterwarnings('ignore')
from scipy.linalg import solve
from scipy.optimize import root, minimize
from scipy import stats
import matplotlib.pyplot as plt

# ===== 任务1 =====
print("===== 任务1：解线性方程组 =====")
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])
x = solve(A, b)
print("解：x =", x[0], ", y =", x[1])
print("代入验证：3x+y =", 3*x[0] + x[1], " x+2y =", x[0] + 2*x[1])
print()

# ===== 任务2 =====
print("===== 任务2：求方程根 =====")
f = lambda x: x**2 - 2
res_pos = root(f, 1)
res_neg = root(f, -1)
print("正根：", res_pos.x[0], " 是否成功：", res_pos.success, " 函数值：", f(res_pos.x[0]))
print("负根：", res_neg.x[0], " 是否成功：", res_neg.success, " 函数值：", f(res_neg.x[0]))
print()

# ===== 任务3 =====
print("===== 任务3：求函数最小值 =====")
f2 = lambda x: x**2 + x + 2
res_min = minimize(f2, 0)
print("极小点：", res_min.x[0], " 函数值：", res_min.fun)
print("解析结果：x =", -1/2, " f(x) =", (-1/2)**2 + (-1/2) + 2)
print("说明：f(x)=x²+x+2的极小点在x=-b/2a=-0.5处，与数值求解结果一致")
print()

# ===== 任务4 =====
print("===== 任务4：独立样本t检验 =====")
np.random.seed(42)
group1 = np.random.normal(0, 1, 100)
group2 = np.random.normal(1, 1, 100)
t_stat, p_value = stats.ttest_ind(group1, group2)
print("t统计量：", t_stat)
print("p值：", p_value)
print("判断：", "p<0.05，拒绝原假设，两组均值有显著差异" if p_value < 0.05 else "p>=0.05，不能拒绝原假设")
print("说明：独立样本t检验比较两组均值是否存在显著差异，p值小于0.05认为差异显著")
print()

# ===== 任务5 =====
print("===== 任务5：Matplotlib可视化 =====")
scores = np.random.normal(70, 10, 1000)
scores = np.clip(scores, 0, 100)
plt.figure()
plt.hist(scores, bins=30)
plt.title("成绩分布直方图")
plt.xlabel("成绩")
plt.ylabel("频数")
plt.show()

plt.figure()
plt.boxplot(scores)
plt.title("成绩箱线图")
plt.ylabel("成绩")
plt.show()

x = np.linspace(0, 5, 100)
plt.figure()
plt.plot(x, x, label="y = x")
plt.plot(x, x**2, label="y = x²")
plt.plot(x, x**3, label="y = x³")
plt.title("函数折线图")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
print("说明：直方图展示成绩分布，箱线图展示中位数和离群点，折线图对比三个函数曲线")
```

```python
import pandas as pd

# ===== 任务1 =====
print("===== 任务1：创建DataFrame并计算总分平均分 =====")
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(records)
df = df.set_index("Name")
df["Total"] = df[["Math", "English", "Python"]].sum(axis=1)
df["Average"] = df[["Math", "English", "Python"]].mean(axis=1)
print("行索引：", df.index.tolist())
print("列名：", df.columns.tolist())
print("前3行：\n", df.head(3))
print("基本统计信息：\n", df.describe())
print()

# ===== 任务2 =====
print("===== 任务2：loc/iloc索引与条件筛选 =====")
print("loc读取Carol：\n", df.loc["Carol"])
carol_pos = df.index.get_loc("Carol")
print("iloc读取Carol（第", carol_pos, "行）：\n", df.iloc[carol_pos])
print("平均分不低于80的学生：\n", df[df["Average"] >= 80])
print()

# ===== 任务3 =====
print("===== 任务3：行列均值计算 =====")
print("三科列均值：\n", df[["Math", "English", "Python"]].mean(axis=0))
print("每名学生行均值：\n", df[["Math", "English", "Python"]].mean(axis=1))
print()

# ===== 任务4 =====
print("===== 任务4：apply/agg/rank =====")
df["MaxScore"] = df[["Math", "English", "Python"]].apply(lambda row: row.max(), axis=1)
print("每名学生最高分：\n", df["MaxScore"])
print("每科最小值/均值/最大值：\n", df[["Math", "English", "Python"]].agg(["min", "mean", "max"]))
print("按Math成绩排名：\n", df["Math"].rank(ascending=False))
print()

# ===== 任务5 =====
print("===== 任务5：分组统计与transform =====")
print("每组人数：\n", df.groupby("Group").size())
print("每组三科平均分：\n", df.groupby("Group")[["Math", "English", "Python"]].mean())
df["GroupPythonMean"] = df.groupby("Group")["Python"].transform("mean")
print("增加组Python平均分后：\n", df[["Group", "Python", "GroupPythonMean"]])
print()

# ===== 任务6 =====
print("===== 任务6：filter筛选组 =====")
filtered = df.groupby("Group").filter(lambda g: g["Python"].mean() >= 80)
print("保留的组和学生：\n", filtered)
print()

# ===== 任务7 =====
print("===== 任务7：协方差与相关系数 =====")
scores = df[["Math", "English", "Python"]]
print("协方差矩阵：\n", scores.cov())
print("皮尔森相关系数矩阵：\n", scores.corr())
print("说明：协方差为正表示两科成绩同向变化，相关系数越接近1线性相关性越强")
```
